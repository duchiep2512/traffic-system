```python
# Cài nếu chưa có (chạy một lần ở terminal/notebook):
# pip install --upgrade google-generativeai

import os, google.generativeai as gen

# Đặt API key (lưu ý: phải là key free của bạn)
os.environ["GOOGLE_API_KEY"] = "<YOUR_API_KEY>"

gen.configure(api_key=os.environ["GOOGLE_API_KEY"])

# Liệt kê các model sinh văn bản có sẵn
models = [m for m in gen.list_models() if "generateContent" in m.supported_generation_methods]
for m in models:
    print(m.name)

# Chọn một model free-tier (thường: "gemini-1.5-flash" hoặc "gemini-1.5-flash-8b")
model_name = "gemini-1.5-flash"

model = gen.GenerativeModel(model_name)
resp = model.generate_content("Tóm tắt tình hình giao thông giờ cao điểm ở Hà Nội trong 3 câu.")
print(resp.text)
```
